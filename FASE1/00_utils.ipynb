```python
def to_numeric_comma(column):
    return column.str.replace(',', '').astype('float')

def remove_contenitori(df_programmi):
    w = (
        Window
        .partitionBy("Data", 'Editore', 'Canale')        # GROUP BY DATE
        .orderBy("ORA_INIZIO_TRX")
        .rowsBetween(1, Window.unboundedFollowing)
    )

    df_flagged = df_programmi.withColumn(
        "min_future_start",
        F.min("ORA_INIZIO_TRX").over(w)
    )

    df_programmi = df_flagged.filter(
        (F.col("min_future_start").isNull()) |
        (F.col("ORA_FINE_TRX") <= F.col("min_future_start"))
    ).drop("min_future_start")

    return df_programmi

def convert_columns_to_numeric(df_programmi):
    for column in COLUMNS_TO_NUMERIC_COMMA:
        df_programmi = df_programmi.withColumn(
            column,
            F.regexp_replace(F.col(column), ',', '').cast('float')
        )

    for column in COLUMNS_TO_NUMERIC:
        df_programmi = df_programmi.withColumn(
            column,
            F.col(column).cast('float')
        )
    
    return df_programmi

def groupby_single_row(df_programmi):
    # add minute_delta column for single total audience computation
    df_programmi = df_programmi.withColumn('Minuto_delta', F.col('Minuto') - F.lag(F.col('Minuto')).over(Window.partitionBy(['Editore', 'Canale', 'Programma', 'Data', 'ORA_INIZIO_TRX', 'ORA_FINE_TRX']).orderBy(F.col('Minuto')))).dropna(subset=['Minuto_delta']).where(F.col('Minuto_delta') > 0)

    # compute weighted audience sums
    w = Window.partitionBy(*['Editore', 'Canale', 'Programma', 'Data', 'ORA_INIZIO_TRX', 'ORA_FINE_TRX'])
    for c in COLUMNS_AUDIENCE_USED:
        df_programmi = df_programmi.withColumn(
            c,
            F.sum(
                F.when(
                    F.col(c).isNotNull(),
                    F.col(c) * F.col("Minuto_delta") / 60
                )
            ).over(w)
            /
            F.sum(
                F.when(
                    F.col(c).isNotNull(),
                    F.col("Minuto_delta") / 60
                )
            ).over(w)
        )
    df_programmi = df_programmi.drop(*['Minuto_delta'])

    # group by single row per programma
    group_columns = [
        'Editore', 'Canale', 'Programma', 'Data',
        'ORA_INIZIO_TRX', 'ORA_FINE_TRX'
    ] + COLUMNS_AUDIENCE_USED

    columns_a = [
        'Cod_Type_Trasm_Int',
        'Cod_Genere_Int',
        'ORA_INIZIO_TRX',
        'ORA_FINE_TRX',
        'Tipo_Trasmissione',
        'Direzione_Di_Genere'
    ]

    # columns_b = ['FlgPrimaVisione']

    w_count = Window.partitionBy(*(list(set(group_columns + columns_a))))
    w_rank  = Window.partitionBy(*group_columns)

    df_ranked = df_programmi

    for c in columns_a:
        if c in ['ORA_INIZIO_TRX', 'ORA_FINE_TRX']:
            continue
        df_ranked = df_ranked.withColumn(
            f"{c}_cnt",
            F.count("*").over(Window.partitionBy(*group_columns, c))
        )

    df_ranked = df_ranked.withColumn(
        "rn",
        F.row_number().over(
            w_rank.orderBy(*[F.col(f"{c}_cnt").desc() for c in columns_a if c not in ['ORA_INIZIO_TRX', 'ORA_FINE_TRX']])
        )
    )

    # for c in columns_b:
    #     df_ranked = df_ranked.withColumn(
    #         c,
    #         F.max(F.col(c).cast("int")).over(w_rank).cast("boolean")
    #     )

    df_programmi = (
        df_ranked
        .filter(F.col("rn") == 1)
        # .select(*group_columns, *columns_a, *columns_b)
        .select(*list(set(group_columns + columns_a)))
    )

    return df_programmi

def convert_time_to_seconds(df_programmi, columns=['Minuto', 'ORA_INIZIO_TRX', 'ORA_FINE_TRX']):
    for column in columns:
        parts = F.split(F.col(column), ":")

        df_programmi = df_programmi.withColumn(
            column,
            F.when(F.size(parts) == 3,
                F.split(F.col(column).cast('string'), ":")[0].cast("int") * 3600 +
                F.split(F.col(column).cast('string'), ":")[1].cast("int") * 60 +
                (F.split(F.col(column).cast('string'), ":")[2].cast("int") if column == 'Minuto' else 0)
            ).when(F.size(parts) == 2,
                F.split(F.col(column).cast('string'), ":")[0].cast("int") * 3600 +
                F.split(F.col(column).cast('string'), ":")[1].cast("int") * 60
            )
        )

    return df_programmi

def compute_total_share(df_programmi):
    # Each program instance is identified by (date, channel, program, start_time, end_time)
    df_programmi = df_programmi.fillna({'LiveVOSDAL': 0.0})
    
    df1 = df_programmi.alias("a")
    df2 = df_programmi.alias("b")

    # Step 1: Join overlapping programs on the same date but different channels
    overlaps = (
        df1.join(
            df2,
            on=(
                (F.col("a.Data") == F.col("b.Data")) & # stessa data
                (F.col("a.Canale") != F.col("b.Canale")) & # canali diversi
                (F.col("b.ORA_INIZIO_TRX") <= F.col("a.ORA_FINE_TRX")) & # b inizia prima della fine di a
                (F.col("b.ORA_FINE_TRX") >= F.col("a.ORA_INIZIO_TRX")) # b finisce dopo l'inizio di a
            ),
            how="left"
        )
        .withColumn(
            "overlap_seconds", # calcola secondi sovrapposti
            F.least(F.col("a.ORA_FINE_TRX"), F.col("b.ORA_FINE_TRX")) -
            F.greatest(F.col("a.ORA_INIZIO_TRX"), F.col("b.ORA_INIZIO_TRX"))
        )
        .filter(F.col("overlap_seconds") > 0)
    )

    weighted = (
        overlaps
        .groupBy('a.Data','a.Programma','a.Canale','a.ORA_INIZIO_TRX','a.ORA_FINE_TRX','a.LiveVOSDAL','b.Canale')
        .agg(
            (F.sum(F.col('b.LiveVOSDAL') * F.col('overlap_seconds')) / F.sum(F.col("overlap_seconds"))).alias('weighted_audience')
        )
    )

    # Step 3: Sum weighted audience across all competing channels for this program instance
    total_weighted = (
        weighted
        .groupBy("a.Data", "a.Programma", "a.Canale", "a.ORA_INIZIO_TRX", "a.ORA_FINE_TRX")
        .agg(F.sum("weighted_audience").alias("total_competing_audience"), F.mean('a.LiveVOSDAL').alias('LiveVOSDAL'))
    )

    # Step 4: Compute share percentage
    result = total_weighted.withColumn(
        "Share",
        F.col("LiveVOSDAL") / (F.col("LiveVOSDAL") + F.col("total_competing_audience"))
    ).select(
        F.col("a.Data").alias("Data"),
        F.col("a.Programma").alias("Programma"),
        F.col("a.Canale").alias("Canale"),
        F.col("a.ORA_INIZIO_TRX").alias("ORA_INIZIO_TRX"),
        F.col("a.ORA_FINE_TRX").alias("ORA_FINE_TRX"),
        # F.col('a.LiveVOSDAL').alias('LiveVOSDAL'),
        "Share",
        F.col('total_competing_audience').alias('LiveVOSDAL_Competition'),
    )

    df_programmi = (
        df_programmi.join(
            result,
            on=[
                "Data",
                "Canale",
                "Programma",
                "ORA_INIZIO_TRX",
                "ORA_FINE_TRX",
                # 'LiveVOSDAL'
            ],
            how="left"
        )
    )

    return df_programmi


def compute_audience_category_percentage(df_programmi):
    uomini_donne_first_screen = ['1st_Screen_LiveVOSDAL_Uomini', '1st_Screen_LiveVOSDAL_Donne']
    maschi_femmine_first_screen = ['1st_Screen_LiveVOSDAL_Maschi', '1st_Screen_LiveVOSDAL_Femmine',]
    maschi_femmine_second_screen = ['2nd_Screen_LiveVOSDAL_Maschi', '2nd_Screen_LiveVOSDAL_Femmine',]
    maschi_first_screen = ['1st_Screen_LiveVOSDAL_Maschi_4_7', '1st_Screen_LiveVOSDAL_Maschi_8_14', '1st_Screen_LiveVOSDAL_Maschi_15_24', '1st_Screen_LiveVOSDAL_Maschi_25_34', '1st_Screen_LiveVOSDAL_Maschi_35_44', '1st_Screen_LiveVOSDAL_Maschi_45_54', '1st_Screen_LiveVOSDAL_Maschi_55_64', '1st_Screen_LiveVOSDAL_Maschi_65_69', '1st_Screen_LiveVOSDAL_Maschi_70_74', '1st_Screen_LiveVOSDAL_Maschi_75plus',]
    femmine_first_screen = ['1st_Screen_LiveVOSDAL_Femmine_4_7', '1st_Screen_LiveVOSDAL_Femmine_8_14', '1st_Screen_LiveVOSDAL_Femmine_15_24', '1st_Screen_LiveVOSDAL_Femmine_25_34', '1st_Screen_LiveVOSDAL_Femmine_35_44', '1st_Screen_LiveVOSDAL_Femmine_45_54', '1st_Screen_LiveVOSDAL_Femmine_55_64', '1st_Screen_LiveVOSDAL_Femmine_65_69', '1st_Screen_LiveVOSDAL_Femmine_70_74', '1st_Screen_LiveVOSDAL_Femmine_75plus',]
    adulti_first_screen = ['1st_Screen_LiveVOSDAL_Bambini_8_14','1st_Screen_LiveVOSDAL_Adulti_15_24', '1st_Screen_LiveVOSDAL_Adulti_25_34', '1st_Screen_LiveVOSDAL_Adulti_35_44', '1st_Screen_LiveVOSDAL_Adulti_45_54', '1st_Screen_LiveVOSDAL_Adulti_55_64', '1st_Screen_LiveVOSDAL_Adulti_65_69', '1st_Screen_LiveVOSDAL_Adulti_70_74', '1st_Screen_LiveVOSDAL_Adulti_75plus',]
    first_second_screen = ['1st_Screen_LiveVOSDAL','2nd_Screen_LiveVOSDAL',]
    second_screen_eta = ['2nd_Screen_LiveVOSDAL_Eta_4_7', '2nd_Screen_LiveVOSDAL_Eta_8_14', '2nd_Screen_LiveVOSDAL_Eta_15_24', '2nd_Screen_LiveVOSDAL_Eta_25_34', '2nd_Screen_LiveVOSDAL_Eta_35_44', '2nd_Screen_LiveVOSDAL_Eta_45_54', '2nd_Screen_LiveVOSDAL_Eta_55_64', '2nd_Screen_LiveVOSDAL_Eta_65plus',]
    device = ['2nd_Screen_LiveVOSDAL_Altri', '2nd_Screen_LiveVOSDAL_Smart_Tv', '2nd_Screen_LiveVOSDAL_Tablet', '2nd_Screen_LiveVOSDAL_Smartphone', '2nd_Screen_LiveVOSDAL_PC',]

    def compute_percentage(df_programmi, columns):
        
        df_programmi = df_programmi.withColumn('summed_columns', sum(F.col(c) for c in columns))

        for column in columns:
            df_programmi = (
                df_programmi
                .withColumn(column + '_perc', F.when(F.col('summed_columns') == 0, F.lit(None)).otherwise(F.col(column) / F.col('summed_columns') * 100))
            )

        for column in columns:
            df_programmi = df_programmi.drop(F.col(column))
            df_programmi = df_programmi.withColumn(column, F.col(column+'_perc'))
            df_programmi = df_programmi.drop(F.col(column+'_perc'))
        
        df_programmi = df_programmi.drop(F.col('summed_columns'))

        return df_programmi

    df_programmi = compute_percentage(df_programmi, uomini_donne_first_screen)
    df_programmi = compute_percentage(df_programmi, maschi_femmine_first_screen)
    df_programmi = compute_percentage(df_programmi, maschi_femmine_second_screen)
    df_programmi = compute_percentage(df_programmi, maschi_first_screen)
    df_programmi = compute_percentage(df_programmi, femmine_first_screen)
    df_programmi = compute_percentage(df_programmi, adulti_first_screen)
    df_programmi = compute_percentage(df_programmi, first_second_screen)
    df_programmi = compute_percentage(df_programmi, second_screen_eta)
    df_programmi = compute_percentage(df_programmi, device)

    return df_programmi
    
```

```python
def read_df_programmi():
    df_programmi = spark.table('ta_coll.whatif.programmi').withColumn('ETA_MEDIA', F.col('ETA_MEDIA').cast('float')).toPandas()

    df_programmi['Data'] = pd.to_datetime(df_programmi['Data'])
    df_programmi['Mese'] = df_programmi['Data'].dt.month
    df_programmi['GiornoSettimana'] = df_programmi['Data'].dt.dayofweek
    df_programmi['Ora'] = np.round(df_programmi['ORA_INIZIO_TRX'] / 3600, decimals = 0) # sec to hour

    return df_programmi

def compute_stats_last_occurrences(group, suffix = '', cutoff_days = 7, cutoff_num_occurrences = 10):
    group = group.sort_values(['Data', 'ORA_INIZIO_TRX'])

    median_values = []
    max_values = []
    min_values = []
    std_values = []
    last_values = []
    uomini_values = []
    share_08_14_values = []
    share_15_24_values = []
    share_25_64_values = []
    share_65_plus_values = []

    for i, row in group.iterrows():
        cutoff_date = row['Data'] - pd.Timedelta(days=cutoff_days)

        past = group.iloc[:i-1]
        past = past.loc[past['Data'] <= cutoff_date]
        past = past.loc[(past['Ora'] >= row['Ora'] - 3) & (past['Ora'] <= row['Ora'] + 3)]
        past = past.tail(cutoff_num_occurrences)
        
        if len(past) == 0:
            median_values.append(None)
            max_values.append(None)
            min_values.append(None)
            std_values.append(None)
            last_values.append(None)
            uomini_values.append(None)
            share_08_14_values.append(None)
            share_15_24_values.append(None)
            share_25_64_values.append(None)
            share_65_plus_values.append(None)
            continue
        
        with warnings.catch_warnings():
            warnings.simplefilter("ignore")
            
            median_values.append(past['Share'].median())
            max_values.append(past['Share'].max())
            min_values.append(past['Share'].min())
            std_values.append(past['Share'].std())
            last_values.append(past['Share'].iloc[-1])
            uomini_values.append(past['1st_Screen_LiveVOSDAL_Uomini'].median())
            share_08_14_values.append(past['1st_Screen_LiveVOSDAL_Bambini_8_14'].median())
            share_15_24_values.append(past['1st_Screen_LiveVOSDAL_Adulti_15_24'].median())
            share_25_64_values.append((past['1st_Screen_LiveVOSDAL_Adulti_25_34'] + past['1st_Screen_LiveVOSDAL_Adulti_35_44'] + past['1st_Screen_LiveVOSDAL_Adulti_45_54'] + past['1st_Screen_LiveVOSDAL_Adulti_55_64']).median())
            share_65_plus_values.append((past['1st_Screen_LiveVOSDAL_Adulti_65_69'] + past['1st_Screen_LiveVOSDAL_Adulti_70_74'] + past['1st_Screen_LiveVOSDAL_Adulti_75plus']).median())
        

    group['StoricoShare' + suffix] = median_values
    group['StoricoShareMax' + suffix] = max_values
    group['StoricoShareMin' + suffix] = min_values
    group['StoricoShareStd' + suffix] = std_values
    group['StoricoShareLast' + suffix] = last_values
    group['StoricoShareUomini' + suffix] = uomini_values
    group['StoricoShare_08_14' + suffix] = share_08_14_values
    group['StoricoShare_15_24' + suffix] = share_15_24_values
    group['StoricoShare_25_64' + suffix] = share_25_64_values
    group['StoricoShare_65_plus' + suffix] = share_65_plus_values

    return group

def compute_stats_prev_next_program(group, suffix = 'Competitor'):
    group = group.sort_values(['Data','ORA_INIZIO_TRX'])

    competitor_columns = [c for c in group.columns if c.endswith(suffix)]

    neighbors = (
        (group['Programma'] == group['Programma'].shift(1)) |
        (group['Programma'] == group['Programma'].shift(2)) |
        (group['Programma'] == group['Programma'].shift(3)) |
        (group['Programma'] == group['Programma'].shift(-1)) |
        (group['Programma'] == group['Programma'].shift(-2)) |
        (group['Programma'] == group['Programma'].shift(-3))
    )

    group['Exists'] = neighbors

    for c in competitor_columns:
        group[c.removesuffix(suffix) + '_prev'] = group[c].shift(1)
        group[c.removesuffix(suffix) + '_next'] = group[c].shift(-1)

        group.loc[group['Exists'], c.removesuffix(suffix) + '_prev'] = group[c].shift(4)
        group.loc[group['Exists'], c.removesuffix(suffix) + '_next'] = group[c].shift(-4)

    group['Programma_prev'] = group['Programma'].shift(1)
    group['Programma_next'] = group['Programma'].shift(-1)

    group.loc[group['Exists'], 'Programma' + '_prev'] = group['Programma'].shift(4)
    group.loc[group['Exists'], 'Programma' + '_next'] = group['Programma'].shift(-4)

    return group

def add_competitor(df_programmi, suffix='', min_competitor_overlap=0.6):
    df1 = df_programmi.alias('a') # main
    df2 = df_programmi.alias('b') # competitor

    overlaps = (
        df1.join(
            df2,
            on = (
                (F.col('a.Data') == F.col('b.Data')) &
                (F.col('a.Canale') != F.col('b.Canale')) &
                (F.col('a.ORA_INIZIO_TRX') <= F.col('b.ORA_FINE_TRX')) &
                (F.col('a.ORA_FINE_TRX') >= F.col('b.ORA_INIZIO_TRX'))
            )
        )
        .withColumn(
            'overlap_seconds',
            F.least(F.col('a.ORA_FINE_TRX'), F.col('b.ORA_FINE_TRX')) - F.greatest(F.col('a.ORA_INIZIO_TRX'), F.col('b.ORA_INIZIO_TRX'))
        ).withColumn('overlap_perc', (F.col('overlap_seconds') / (F.col('a.ORA_FINE_TRX') - F.col('a.ORA_INIZIO_TRX'))))
        .where(F.col('overlap_perc') >= min_competitor_overlap)
    )

    w = Window.partitionBy(
        F.col('a.Canale'),
        F.col('a.Data'),
        F.col('a.Programma'),
        F.col('a.ORA_INIZIO_TRX'),
        F.col('a.ORA_FINE_TRX')
    ).orderBy(F.col(f'b.StoricoShare{suffix}').desc())

    overlaps = overlaps.withColumn(
        'rn', F.row_number().over(w)
    ).filter(F.col('rn') == 1)

    storico_competitor = overlaps.groupBy(
        'a.Canale',
        'a.Data',
        'a.Programma',
        'a.ORA_INIZIO_TRX',
        'a.ORA_FINE_TRX'
    ).agg(
        F.first('b.Programma').alias('Programma_competitor'),
        F.first('b.ORA_INIZIO_TRX').alias('ORA_INIZIO_TRX_competitor'),
        F.first('b.ORA_FINE_TRX').alias('ORA_FINE_TRX_competitor'),
        F.first(f'b.StoricoShare{suffix}').alias('StoricoShare_competitor'),
        F.first(f'b.StoricoShareLast{suffix}').alias('StoricoShareLast_competitor'),
        F.first(f'b.StoricoShare_08_14{suffix}').alias('StoricoShare_08_14_competitor'),
        F.first(f'b.StoricoShare_15_24{suffix}').alias('StoricoShare_15_24_competitor'),
        F.first(f'b.StoricoShare_25_64{suffix}').alias('StoricoShare_25_64_competitor'),
        F.first(f'b.StoricoShare_65_plus{suffix}').alias('StoricoShare_65_plus_competitor'),
        F.first('b.Canale').alias('Canale_competitor'),
        F.first('b.DES_GENERE_ESTESA_INT').alias('DES_GENERE_ESTESA_INT_competitor'),
        F.first('b.DES_GENERE_FILM_INT').alias('DES_GENERE_FILM_INT_competitor'),
        F.first('b.DES_GENERE_SPORT_INT').alias('DES_GENERE_SPORT_INT_competitor'),
        F.first(f'b.StoricoShareUomini{suffix}').alias('StoricoShareUomini_competitor'),
        F.first('overlap_perc').alias('overlap_perc_competitor')
    )

    storico_competitor = (
        storico_competitor.select(
            F.col('a.Canale').alias('Canale'),
            F.col('a.Data').alias('Data'),
            F.col('a.Programma').alias('Programma'),
            F.col('a.ORA_INIZIO_TRX').alias('ORA_INIZIO_TRX'),
            F.col('a.ORA_FINE_TRX').alias('ORA_FINE_TRX'),
            F.col('Programma_competitor'),
            F.col('ORA_INIZIO_TRX_competitor'),
            F.col('ORA_FINE_TRX_competitor'),
            F.col('StoricoShare_competitor'),
            F.col('StoricoShareLast_competitor'),
            F.col('StoricoShareUomini_competitor'),
            F.col('StoricoShare_08_14_competitor'),
            F.col('StoricoShare_15_24_competitor'),
            F.col('StoricoShare_25_64_competitor'),
            F.col('StoricoShare_65_plus_competitor'),
            F.col('Canale_competitor'),
            F.col('DES_GENERE_ESTESA_INT_competitor'),
            F.col('DES_GENERE_FILM_INT_competitor'),
            F.col('DES_GENERE_SPORT_INT_competitor'),
            F.col('overlap_perc_competitor'),
        )
    )

    df_programmi = df_programmi.join(storico_competitor, on=['Canale', 'Data', 'Programma', 'ORA_INIZIO_TRX', 'ORA_FINE_TRX'], how='left')

    return df_programmi
```
